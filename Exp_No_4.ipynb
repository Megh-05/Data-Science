```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

from tensorflow.keras.datasets import cifar10

# Load CIFAR-10 Dataset

(train_images, train_labels), (test_images, test_labels) = cifar10.load_data()

# Normalize Pixel Values

train_images = train_images.astype('float32') / 255
test_images = test_images.astype('float32') / 255

# Dataset Categories

categories = [
    'Plane',
    'Car',
    'Bird',
    'Cat',
    'Deer',
    'Dog',
    'Frog',
    'Horse',
    'Ship',
    'Truck'
]

# Display Random Images

plt.figure(figsize=(9, 9))

for index in range(9):

    plt.subplot(3, 3, index + 1)

    plt.imshow(train_images[index])

    plt.title(categories[int(train_labels[index])])

    plt.xticks([])
    plt.yticks([])

plt.tight_layout()
plt.show()

# Build CNN Architecture

cnn_model = Sequential()

# Convolution Block 1

cnn_model.add(
    Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation='relu',
        padding='same',
        input_shape=(32, 32, 3)
    )
)

cnn_model.add(MaxPooling2D(pool_size=(2, 2)))

# Convolution Block 2

cnn_model.add(
    Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation='relu',
        padding='same'
    )
)

cnn_model.add(MaxPooling2D(pool_size=(2, 2)))

# Convolution Block 3

cnn_model.add(
    Conv2D(
        filters=128,
        kernel_size=(3, 3),
        activation='relu'
    )
)

# Flatten Layer

cnn_model.add(Flatten())

# Fully Connected Layers

cnn_model.add(Dense(128, activation='relu'))

cnn_model.add(Dropout(0.3))

cnn_model.add(Dense(10, activation='softmax'))

# Display Model Structure

cnn_model.summary()

# Compile Neural Network

cnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train CNN Model

training_result = cnn_model.fit(
    train_images,
    train_labels,
    epochs=8,
    batch_size=64,
    validation_split=0.1
)

# Test Model Performance

loss_value, accuracy_value = cnn_model.evaluate(
    test_images,
    test_labels
)

print("\nCNN Test Accuracy :", accuracy_value)

# Accuracy Graph

plt.figure(figsize=(7,5))

plt.plot(
    training_result.history['accuracy'],
    label='Train Accuracy'
)

plt.plot(
    training_result.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()

plt.grid(True)

plt.show()
```
