```python
# =========================
# Simple Embedding Example
# =========================

import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Flatten, Dropout

# Input Data
input_data = np.array([0, 1, 2, 3, 4, 5])
output_data = np.array([1, 0, 1, 0, 1, 0])

# Create Sequential Model
embedding_model = Sequential()

embedding_model.add(
    Embedding(
        input_dim=12,
        output_dim=5,
        input_shape=(1,)
    )
)

embedding_model.add(Flatten())

embedding_model.add(Dense(10, activation='relu'))
embedding_model.add(Dropout(0.1))

embedding_model.add(Dense(1, activation='sigmoid'))

# Compile Model
embedding_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Model
embedding_model.fit(
    input_data,
    output_data,
    epochs=15,
    batch_size=2
)

# Prediction
result = embedding_model.predict(np.array([3]))

print("Prediction Output :", result)
```

```python
# =====================================
# Recommendation System using Embedding
# =====================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Embedding,
    Flatten,
    Dense,
    Multiply
)

from tensorflow.keras.optimizers import Adam

# Create Dataset

dataset = {
    'customer_id': [0,0,0,1,1,1,2,2,2,3,3,3],
    'product_id': [0,1,2,0,1,2,0,1,2,0,1,2],
    'score': [5,4,1,4,2,5,2,5,4,1,4,5]
}

ratings_df = pd.DataFrame(dataset)

# Separate Features and Labels

features = ratings_df[['customer_id', 'product_id']]
labels = ratings_df['score']

# Train Test Split

x_train, x_test, y_train, y_test = train_test_split(
    features,
    labels,
    test_size=0.25,
    random_state=10
)

# Total Users and Products

total_users = ratings_df.customer_id.nunique()
total_products = ratings_df.product_id.nunique()

embedding_units = 32

# User Input Layer

customer_input = Input(shape=(1,))

customer_embedding = Embedding(
    input_dim=total_users,
    output_dim=embedding_units
)(customer_input)

customer_vector = Flatten()(customer_embedding)

# Product Input Layer

product_input = Input(shape=(1,))

product_embedding = Embedding(
    input_dim=total_products,
    output_dim=embedding_units
)(product_input)

product_vector = Flatten()(product_embedding)

# Combine User and Product Vectors

merged_vector = Multiply()([customer_vector, product_vector])

# Hidden Layers

hidden_layer = Dense(64, activation='relu')(merged_vector)

hidden_layer = Dense(32, activation='relu')(hidden_layer)

final_output = Dense(1)(hidden_layer)

# Build Recommendation Model

recommendation_model = Model(
    inputs=[customer_input, product_input],
    outputs=final_output
)

# Compile Model

recommendation_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='mean_squared_error',
    metrics=['mae']
)

# Train Model

recommendation_model.fit(
    [x_train.customer_id, x_train.product_id],
    y_train,
    epochs=15,
    batch_size=2,
    validation_split=0.1
)

# Evaluate Model

test_loss, test_mae = recommendation_model.evaluate(
    [x_test.customer_id, x_test.product_id],
    y_test
)

print("\nTesting Loss :", test_loss)
print("Testing MAE :", test_mae)
```

```python
# ==================================
# Pairwise Ranking Recommendation NN
# ==================================

import tensorflow as tf
import numpy as np

from tensorflow.keras.layers import (
    Input,
    Embedding,
    Flatten,
    Dot,
    Lambda
)

from tensorflow.keras.models import Model

# Parameters

users_count = 6
items_count = 12
vector_size = 10

# Input Layers

user_layer = Input(shape=(1,))
liked_item_layer = Input(shape=(1,))
disliked_item_layer = Input(shape=(1,))

# Embedding Layers

user_embed = Embedding(
    input_dim=users_count,
    output_dim=vector_size
)(user_layer)

liked_embed = Embedding(
    input_dim=items_count,
    output_dim=vector_size
)(liked_item_layer)

disliked_embed = Embedding(
    input_dim=items_count,
    output_dim=vector_size
)(disliked_item_layer)

# Flatten Embeddings

user_output = Flatten()(user_embed)

liked_output = Flatten()(liked_embed)

disliked_output = Flatten()(disliked_embed)

# Similarity Scores

liked_score = Dot(axes=1)(
    [user_output, liked_output]
)

disliked_score = Dot(axes=1)(
    [user_output, disliked_output]
)

# Custom Difference Layer

difference = Lambda(
    lambda x: x[0] - x[1]
)([liked_score, disliked_score])

# Final Model

ranking_model = Model(
    inputs=[
        user_layer,
        liked_item_layer,
        disliked_item_layer
    ],
    outputs=difference
)

# Compile Model

ranking_model.compile(
    optimizer='adam',
    loss='mean_absolute_error'
)

# Dummy Training Data

train_users = np.array([0, 1, 2, 3])

positive_products = np.array([2, 3, 4, 5])

negative_products = np.array([7, 8, 9, 10])

labels = np.ones((4, 1))

# Train Model

ranking_model.fit(
    [train_users, positive_products, negative_products],
    labels,
    epochs=15,
    batch_size=2
)

print("\nRecommendation Ranking Model Trained Successfully")
```
